## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Import required libraries
import pandas as pd
import numpy as np

# Create a copy for cleaning
data = df.copy()

# Check initial shape
print("Initial shape:", data.shape)


# Remove duplicate rows
print("Duplicate rows:", data.duplicated().sum())

data = data.drop_duplicates()
````

````python
# Clean column names
data.columns = (
    data.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("/", "_", regex=False)
    .str.replace("#", "No", regex=False)
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)

print("\nCleaned column names:")
print(data.columns.tolist())
````

````python
# Remove extra spaces from text columns
text_columns = data.select_dtypes(include="object").columns

for col in text_columns:
    data[col] = data[col].str.strip()
````

````python
# Convert date columns to datetime

date_columns = [
    'PQ_First_Sent_to_Client_Date',
    'PO_Sent_to_Vendor_Date',
    'Scheduled_Delivery_Date',
    'Delivered_to_Client_Date',
    'Delivery_Recorded_Date'
]
for col in date_columns:
    if col in data.columns:
        data[col] = pd.to_datetime(data[col], errors='coerce')
````

````python
# Convert numerical columns to numeric

numeric_columns = [
    'Unit_of_Measure_Per_Pack',
    'Line_Item_Quantity',
    'Line_Item_Value',
    'Pack_Price',
    'Unit_Price',
    'Weight_Kilograms',
    'Freight_Cost_USD',
    'Line_Item_Insurance_USD'
]

for col in numeric_columns:
    if col in data.columns:
        data[col] = pd.to_numeric(data[col], errors='coerce')
````

````python
# Check data types
print("\nData types:")
print(data.dtypes)
````

````python
# Delivery delay in days
data['Delivery_Delay_Days'] = (
    data['Delivered_to_Client_Date'] -
    data['Scheduled_Delivery_Date']
).dt.days
````

````python
# Absolute delivery time from PO to delivery
data['Delivery_Time_Days'] = (
    data['Delivered_to_Client_Date'] -
    data['PO_Sent_to_Vendor_Date']
).dt.days
````

````python
# Delivery status
data['Delivery_Status'] = np.where(
    data['Delivery_Delay_Days'] > 0,
    'Delayed',
    'On Time'
)
````

````python
display(
    data[
        ['Scheduled_Delivery_Date',
         'Delivered_to_Client_Date',
         'Delivery_Delay_Days',
         'Delivery_Status']
    ].head()
)
````

````python
# Replace blank strings with NaN
data.replace(r'^\s*$', np.nan, regex=True, inplace=True)
````

````python
# Check missing values
print("\nMissing values:")
print(data.isnull().sum())
````

````python
# For numerical columns
numeric_cols = data.select_dtypes(include=np.number).columns

for col in numeric_cols:
    data[col] = data[col].fillna(data[col].median())
````

````python
# For categorical columns
categorical_cols = data.select_dtypes(include='object').columns

for col in categorical_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
````

````python
data[categorical_cols] = data[categorical_cols].fillna("Unknown")
````

````python
# For date columns
date_cols = data.select_dtypes(include='datetime64[ns]').columns

print(data[date_cols].isnull().sum())
````

````python
missing_report = pd.DataFrame({
    'Missing_Count': data.isna().sum(),
    'Missing_%': (data.isna().sum() / len(data) * 100).round(2)
})

missing_report = missing_report[
    missing_report['Missing_Count'] > 0
].sort_values('Missing_Count', ascending=False)

missing_report
````
